You are supporting an investment team that is reviewing a universe of European financial institutions. The team wants a concise first-pass analysis of profitability, scale and regional differences.

The dataset is **synthetic** and was created for this portfolio project. All institutions and figures are fictional.

```python
# Import pandas and load the dataset
import pandas as pd
financial_institutions = pd.read_csv("financial_institutions.csv")
```

## 2. Identify highly profitable institutions

The investment team considers an institution to have **high profitability** when its ROE is at least **12%**.

Create a DataFrame named `high_roe_institutions` containing the institution name, country, institution type and ROE for institutions meeting this threshold. Sort the result from highest to lowest ROE.

```python
# Your code here
high_roe_institutions = financial_institutions[financial_institutions['roe_pct']>=12.0][['institution_name', 'country', 'institution_type', 'roe_pct']].sort_values('roe_pct', ascending=False)
print(high_roe_institutions)
```

```python
print(financial_institutions[financial_institutions['roe_pct']>=12.0].groupby('region')['institution_name'].count())
print(financial_institutions[financial_institutions['roe_pct']>=12.0].groupby('country')['institution_name'].count())
print(financial_institutions[financial_institutions['roe_pct']>=12.0].groupby('institution_type')['institution_name'].count())
```

## 3. Build a simple scale indicator

Create a new column named `business_scale_bn_eur` equal to:

`assets_bn_eur + aum_bn_eur`

Then identify the **10 institutions with the largest business scale**. Store the result in `top_10_scale` and show institution name, country, institution type and business scale.

```python
# Your code here
financial_institutions['business_scale_bn_eur'] = financial_institutions['assets_bn_eur']+financial_institutions['aum_bn_eur']
top_10_scale = financial_institutions[['institution_name', 'country', 'institution_type', 'business_scale_bn_eur']].sort_values('business_scale_bn_eur', ascending=False).head(10)
print(top_10_scale)
```

## 4. Compare profitability across regions

Group the institutions by `region` and calculate the **count**, **mean**, and **standard deviation** of `roe_pct`.

Store the result in `roe_by_region` and rename the resulting columns to:

- `institution_count`
- `mean_roe`
- `std_roe`

```python
# Your code here
regions_infos = financial_institutions.groupby('region')['roe_pct'].agg(['count', 'mean', 'std'])
roe_by_region = pd.DataFrame({
    "institution_count": regions_infos['count'],
    "mean_roe": regions_infos['mean'].round(2),
    "std_roe": regions_infos['std'].round(2)
})
print(roe_by_region)
```

## 5. Find the region with the greatest ROE dispersion

Using the table created above, identify the region with the **highest standard deviation of ROE**.

Create a one-row DataFrame named `largest_roe_dispersion` containing that region and its three summary statistics.

```python
# Your code here
largest_roe_dispersion = roe_by_region[roe_by_region['std_roe']==roe_by_region['std_roe'].max()]
print(largest_roe_dispersion)
```
